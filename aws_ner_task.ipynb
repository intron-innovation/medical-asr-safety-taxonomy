```python
import boto3

# Create client
client = boto3.client(
    service_name='comprehendmedical',
    region_name='us-east-1'
)

# Input clinical text
text = "Patient takes aspirin 100 mg daily for headache"

# Call API
response = client.detect_entities_v2(Text=text)

# Print results
for entity in response['Entities']:
    print("Text:", entity['Text'])
    print("Category:", entity['Category'])
    print("Type:", entity['Type'])
    print("Score:", entity['Score'])
    
    # Attributes (dosage, frequency, etc.)
    if 'Attributes' in entity:
        for attr in entity['Attributes']:
            print("  Attribute:", attr['Type'], "-", attr['Text'])
    
    print("------")
```
